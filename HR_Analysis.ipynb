```python
#Handle Missing Values
df[df['YearsWithCurrManager'].isnull()]
```

```python
median_value = df['YearsWithCurrManager'].median()

print("Median YearsWithCurrManager:", median_value)
```

```python
#Fill the missing values:

df['YearsWithCurrManager'] = df['YearsWithCurrManager'].fillna(median_value)
```

```python
Filtering
```

```python
#Employees who left the company
attrition_yes = df[df['Attrition'] == 'Yes']

print("Employees who left:", len(attrition_yes))

attrition_yes.head()
```

```python
#Employees working overtime
overtime_employees = df[df['OverTime'] == 'Yes']

print("Employees working overtime:", len(overtime_employees))

overtime_employees.head()
```

```python
#Employees with high income
high_income = df[df['MonthlyIncome'] > 10000]

print("Employees earning above 10,000:", len(high_income))

high_income.head()
```

```python
#Employees living far from workplace
far_employees = df[df['DistanceFromHome'] > 10]

print("Employees living more than 10 km away:", len(far_employees))

far_employees.head()
```

```python
Aggregation
#groupby(), mean(), count(), sum(), etc.
```

```python
#Employees by Department
department_count = df.groupby('Department').size()

print(department_count)
```

```python
#Average Monthly Income by Department
avg_income_department = df.groupby('Department')['MonthlyIncome'].mean()

print(avg_income_department)
```

```python
#Average Income by Job Role
avg_income_role = (
    df.groupby('JobRole')['MonthlyIncome']
    .mean()
    .sort_values(ascending=False)
)

print(avg_income_role)
```

```python
#Attrition by Department
attrition_department = pd.crosstab(
    df['Department'],
    df['Attrition']
)

print(attrition_department)
```

```python
#Attrition percentage
attrition_rate = (
    df.groupby('Department')['Attrition']
    .apply(lambda x: (x == 'Yes').mean() * 100)
    .sort_values(ascending=False)
)

print(attrition_rate)
```

```python
#Overtime vs Attrition
overtime_attrition = pd.crosstab(
    df['OverTime'],
    df['Attrition']
)

print(overtime_attrition)
```

```python
#Average job satisfaction by department
satisfaction_department = (
    df.groupby('Department')['JobSatisfaction']
    .mean()
    .sort_values(ascending=False)
)

print(satisfaction_department)
```

```python
EDA — Exploratory Data Analysis
```

```python
#Overall Attrition
attrition_count = df['Attrition'].value_counts()

print(attrition_count)
```

```python
#Attrition by Age Group
agegroup_attrition = pd.crosstab(
    df['AgeGroup'],
    df['Attrition']
)

agegroup_attrition
```

```python
#Attrition by Job Role
jobrole_attrition = pd.crosstab(
    df['JobRole'],
    df['Attrition']
)

jobrole_attrition
```

```python
Data Visualization
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
#Attrition Distribution
plt.figure(figsize=(6,4))

sns.countplot(data=df, x='Attrition')

plt.title('Employee Attrition Distribution')
plt.xlabel('Attrition')
plt.ylabel('Number of Employees')

plt.show()
```

```python
#Employees by Department
plt.figure(figsize=(8,5))

sns.countplot(data=df, x='Department')

plt.title('Employees by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')

plt.xticks(rotation=20)

plt.show()
```

```python
#Attrition by Department
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x='Department',
    hue='Attrition'
)

plt.title('Employee Attrition by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')

plt.show()
```

```python
#Overtime vs Attrition
plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x='OverTime',
    hue='Attrition'
)

plt.title('Overtime vs Employee Attrition')
plt.xlabel('Overtime')
plt.ylabel('Number of Employees')

plt.show()
```

```python
#Monthly Income Distribution
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='MonthlyIncome',
    bins=30,
    kde=True
)

plt.title('Monthly Income Distribution')
plt.xlabel('Monthly Income')
plt.ylabel('Number of Employees')

plt.show()
```

```python
#Age Distribution
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='Age',
    bins=20,
    kde=True
)

plt.title('Employee Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Employees')

plt.show()
```

```python
#Age vs Monthly Income
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='Age',
    y='MonthlyIncome',
    hue='Attrition'
)

plt.title('Age vs Monthly Income')
plt.xlabel('Age')
plt.ylabel('Monthly Income')

plt.show()
```

```python
#Correlation Heatmap
plt.figure(figsize=(14,10))

correlation = df.select_dtypes(include=np.number).corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f'
)

plt.title('Correlation Heatmap of HR Metrics')

plt.show()
```

### Project Overview Summary

This project involved a comprehensive HR Analytics dataset analysis. The initial steps included:

*   **Data Loading and Initial Inspection**: The `HR_Analytics.csv` dataset was loaded into a Pandas DataFrame. The `.head()`, `.shape`, `.columns`, `.info()`, and `.describe()` methods were used to get a preliminary understanding of the data's structure, column names, data types, and basic statistics.
*   **Missing Values and Duplicates**: The dataset was checked for missing values and duplicate records. It was found that 'YearsWithCurrManager' had 57 missing values, and there were 7 duplicate rows.
*   **Irrelevant Column Removal**: Several columns identified as irrelevant for the analysis, such as 'EmpID', 'EmployeeNumber', 'EmployeeCount', 'Over18', and 'StandardHours', were removed to streamline the dataset.

### Data Preprocessing & Cleaning

Following the initial inspection and removal of irrelevant columns, the dataset underwent further cleaning:

*   **Missing Value Imputation**: The 'YearsWithCurrManager' column, which had 57 missing values, was imputed using its median value (3.0) to ensure data completeness. A final check confirmed no remaining missing values.
*   **Data Type Verification**: Data types were checked to ensure consistency and suitability for analysis. All numerical columns were correctly identified, and categorical columns were suitable for their intended use.
*   **Unique Value Review**: Unique values for each column were reviewed to identify any inconsistencies or anomalies not caught by earlier checks.

### Filtering & Aggregation

After cleaning, the data was filtered and aggregated to extract specific insights:

*   **Filtering**: The dataset was filtered to identify specific employee groups, including:
    *   Employees who had left the company (`Attrition == 'Yes'`).
    *   Employees working overtime (`OverTime == 'Yes'`).
    *   Employees with high monthly income (`MonthlyIncome > 10000`).
    *   Employees living far from the workplace (`DistanceFromHome > 10`).
*   **Aggregation**: Various aggregations were performed to understand group-wise trends:
    *   Number of employees by `Department`.
    *   Average `MonthlyIncome` by `Department` and `JobRole`.
    *   `Attrition` rates by `Department` and `OverTime` status.
    *   Average `JobSatisfaction` by `Department`.

*   **Income and Age Distribution**: Plotted histograms to show the distribution of 'MonthlyIncome' and 'Age'.
